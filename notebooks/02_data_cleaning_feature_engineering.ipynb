```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Admin\OneDrive\Desktop\climate-intelligence-prototype\data\climate_data.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())

# Check duplicate rows
duplicate_count = df.duplicated().sum()

print("\nTotal duplicate rows:", duplicate_count)
```

```python
# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print("Date datatype:", df["Date"].dtype)

print("\nInvalid dates after conversion:", df["Date"].isnull().sum())
```

```python
# Check for logically invalid values

print("Humidity values outside 0-100:")
print(((df["Humidity (%)"] < 0) | (df["Humidity (%)"] > 100)).sum())

print("\nCloud cover values outside 0-100:")
print(((df["Cloud_Cover (%)"] < 0) | (df["Cloud_Cover (%)"] > 100)).sum())

print("\nNegative rainfall values:")
print((df["Rainfall (mm)"] < 0).sum())

print("\nNegative wind speed values:")
print((df["Wind_Speed (km/h)"] < 0).sum())

print("\nInvalid temperature relationship (Max < Min):")
print((df["Temperature_Max (°C)"] < df["Temperature_Min (°C)"]).sum())

print("\nNegative AQI values:")
print((df["AQI"] < 0).sum())

print("\nInvalid pressure values (<= 0):")
print((df["Pressure (hPa)"] <= 0).sum())
```

```python
# Check outliers using the IQR method

numeric_columns = [
    "Temperature_Max (°C)",
    "Temperature_Min (°C)",
    "Temperature_Avg (°C)",
    "Humidity (%)",
    "Rainfall (mm)",
    "Wind_Speed (km/h)",
    "AQI",
    "Pressure (hPa)",
    "Cloud_Cover (%)"
]

outlier_summary = {}

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[column] < lower_bound) | (df[column] > upper_bound)).sum()

    outlier_summary[column] = outliers

print("Outlier count for each numerical column:")
for column, count in outlier_summary.items():
    print(f"{column}: {count}")
```

```python
Q1 = df["Rainfall (mm)"].quantile(0.25)
Q3 = df["Rainfall (mm)"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nRainfall statistics:")
print(df["Rainfall (mm)"].describe())

print("\nMaximum rainfall:", df["Rainfall (mm)"].max())
```

```python
# Create time-based features from Date

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Day_of_Week"] = df["Date"].dt.dayofweek
df["Day_Name"] = df["Date"].dt.day_name()

df[["Date", "Year", "Month", "Day", "Day_of_Week", "Day_Name"]].head()
```

```python
# Create weather-based derived features

# Temperature difference between maximum and minimum
df["Temperature_Range (°C)"] = (
    df["Temperature_Max (°C)"] - df["Temperature_Min (°C)"]
)

# Average temperature difference from maximum temperature
df["Max_Avg_Temp_Difference (°C)"] = (
    df["Temperature_Max (°C)"] - df["Temperature_Avg (°C)"]
)

# Average temperature difference from minimum temperature
df["Avg_Min_Temp_Difference (°C)"] = (
    df["Temperature_Avg (°C)"] - df["Temperature_Min (°C)"]
)

# Combined humidity and temperature indicator
df["Humidity_Temperature_Index"] = (
    df["Humidity (%)"] * df["Temperature_Avg (°C)"]
)

df[
    [
        "Temperature_Max (°C)",
        "Temperature_Min (°C)",
        "Temperature_Avg (°C)",
        "Temperature_Range (°C)",
        "Max_Avg_Temp_Difference (°C)",
        "Avg_Min_Temp_Difference (°C)",
        "Humidity_Temperature_Index"
    ]
].head()
```

```python
# Check correlations among numerical features

correlation_matrix = df.select_dtypes(include=["number"]).corr()

correlation_matrix.round(2)
```

```python
# Remove redundant temperature difference features

df.drop(
    columns=[
        "Max_Avg_Temp_Difference (°C)",
        "Avg_Min_Temp_Difference (°C)"
    ],
    inplace=True
)

print("Removed redundant features:")
print("Max_Avg_Temp_Difference (°C)")
print("Avg_Min_Temp_Difference (°C)")

print("\nRemaining columns:")
print(df.columns.tolist())
```

```python
# Final validation of the cleaned dataset

print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())

print("\nFinal columns:")
print(df.columns.tolist())
```

```python
# Save the cleaned and feature-engineered dataset

output_path = "../data/processed/climate_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_path)
```

# Summary

## Data Cleaning

- The original climate dataset was loaded without modifying the original file.
- Dataset shape was verified as 7310 rows and 13 columns.

- No duplicate rows were found.
- Date was converted from string format to datetime format.

The following features were created from the Date column:

- Year
- Month
- Day
- Day_of_Week
- Day_Name

The following weather-based features were created:

- Temperature_Range (°C)

Highly redundant temperature-difference features were removed based on correlation analysis.

AQI was retained as the target variable. AQI was not used to create any derived feature in order to avoid data leakage.
